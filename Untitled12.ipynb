```python
import streamlit as st
import pandas as pd
import plotly.express as px

# --- CONFIGURAZIONE DELLA PAGINA ---
st.set_page_config(
    page_title="Mappa Emissioni CO₂",
    page_icon="🌍",
    layout="wide"
)

# --- FUNZIONE PER SCARICARE I DATI (WEB SCRAPING/FETCHING) ---
@st.cache_data
def load_data():
    # URL diretto al dataset Raw (formato CSV)
    # Usiamo OWID perché ElectricityMaps richiede login/API key privata
    url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"

    # Legge il CSV direttamente dal link (scraping dei dati strutturati)
    df = pd.read_csv(url)

    # Filtriamo le colonne essenziali per ridurre il carico
    cols_to_keep = ['iso_code', 'country', 'year', 'co2_per_capita', 'co2', 'gdp', 'population']
    df = df[cols_to_keep]

    # Rimuoviamo righe che non sono paesi (es. 'World', 'Europe') basandoci su iso_code
    df = df[df['iso_code'].notna()]

    return df

# --- INTERFACCIA UTENTE ---
st.title("🌍 Mappa Interattiva Emissioni CO₂")
st.markdown("Questa applicazione scarica i dati pubblici sulle emissioni e genera una visualizzazione interattiva.")

# Caricamento dati
with st.spinner('Scaricamento dei dati in corso...'):
    try:
        df = load_data()

        # Slider per selezionare l'anno
        min_year = int(df['year'].min())
        max_year = int(df['year'].max())
        selected_year = st.slider("Seleziona l'anno", min_year, max_year, 2022)

        # Filtro dati per l'anno selezionato
        df_year = df[df['year'] == selected_year]

        # --- CREAZIONE DEL GRAFICO PLOTLY ---
        st.subheader(f"Emissioni CO₂ pro capite nel {selected_year}")

        fig = px.choropleth(
            df_year,
            locations="iso_code",       # Codice ISO per mappare i paesi
            color="co2_per_capita",     # Colore basato sulle emissioni pro capite
            hover_name="country",       # Nome paese al passaggio del mouse
            color_continuous_scale=px.colors.sequential.Plasma,
            range_color=(0, 20),        # Fissa la scala per evitare sbalzi di colore
            labels={'co2_per_capita': 'Tonnellate per persona'},
            title=f"Emissioni CO₂ Pro Capite ({selected_year})"
        )

        # Miglioramenti estetici della mappa
        fig.update_geos(
            showframe=False,
            showcoastlines=True,
            projection_type="natural earth"
        )
        fig.update_layout(
            margin={"r":0,"t":50,"l":0,"b":0},
            height=600
        )

        # Visualizza il grafico
        st.plotly_chart(fig, use_container_width=True)

        # Mostra i dati grezzi in una tabella espandibile
        with st.expander("Visualizza dati grezzi per l'anno selezionato"):
            st.dataframe(df_year.sort_values(by="co2_per_capita", ascending=False))

    except Exception as e:
        st.error(f"Errore nel caricamento dei dati: {e}")

# Footer
st.markdown("---")
st.markdown("Dati forniti da *Our World in Data* (Repository GitHub). Sviluppato con Streamlit e Plotly.")
```

```python
# 1. Installa le librerie (se non presenti)
!pip install -q pandas plotly ipywidgets

import pandas as pd
import plotly.express as px
import plotly.io as pio
from ipywidgets import interact, widgets
from IPython.display import display, clear_output

# --- IMPORTANTE PER COLAB ---
# Forza il renderer su colab per assicurarsi che i grafici appaiano
pio.renderers.default = 'colab'

# --- CARICAMENTO DATI ---
print("Caricamento dati in corso...")
url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
df = pd.read_csv(url)

# Filtro e pulizia
cols = ['iso_code', 'country', 'year', 'co2_per_capita']
df = df[cols]
df = df[df['iso_code'].notna()]
print("Dati caricati! Generazione del grafico...")

# --- CREAZIONE DEL CONTENITORE ---
# Creiamo un "foglio bianco" dove disegneremo il grafico
output_container = widgets.Output()

# --- FUNZIONE DI AGGIORNAMENTO ---
def aggiorna_mappa(anno):
    # 1. Filtra i dati
    df_anno = df[df['year'] == anno]

    # 2. "Entra" nel contenitore per disegnare
    with output_container:
        # Pulisce il grafico precedente per non sovrapporli
        clear_output(wait=True)

        if df_anno.empty:
            print(f"Nessun dato per l'anno {anno}")
            return

        # 3. Crea il grafico Plotly
        fig = px.choropleth(
            df_anno,
            locations="iso_code",
            color="co2_per_capita",
            hover_name="country",
            color_continuous_scale=px.colors.sequential.Plasma,
            range_color=(0, 20),
            title=f"Emissioni CO₂ Pro Capite nel {anno}",
            labels={'co2_per_capita': 'Tons/persona'}
        )

        # Impostazioni grafiche (margini ridotti)
        fig.update_layout(
            margin={"r":0,"t":50,"l":0,"b":0},
            height=500
        )

        # 4. MOSTRA IL GRAFICO
        fig.show()

# --- VISUALIZZAZIONE FINALE ---
# Creiamo lo slider
slider = widgets.IntSlider(
    min=1950,
    max=int(df['year'].max()),
    step=1,
    value=2022,
    description='Anno:',
    continuous_update=False # Evita scatti mentre trascini
)

# Colleghiamo lo slider alla funzione
ui = widgets.VBox([slider, output_container])

# Inizializziamo il grafico con l'anno di partenza
aggiorna_mappa(2022)

# Colleghiamo il movimento dello slider alla funzione
slider.observe(lambda change: aggiorna_mappa(change['new']), names='value')

# Mostriamo tutto
display(ui)
```

```python
# 1. Installazione librerie (se necessario)
!pip install -q pandas plotly ipywidgets

import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
from ipywidgets import interact, widgets

# --- CARICAMENTO DATI ---
print("Scaricamento dati...")
url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
df = pd.read_csv(url)

# Filtro colonne e pulizia
df = df[['iso_code', 'country', 'year', 'co2_per_capita']]
df = df[df['iso_code'].notna()] # Rimuove continenti/regioni
print("Dati pronti.")

# --- 1. CREIAMO LA MAPPA INIZIALE (Base) ---
# Prendiamo un anno di riferimento per creare la struttura
year_start = 2022
df_start = df[df['year'] == year_start]

# Creiamo la figura usando Graph Objects (più stabile di Express per i widget)
fig = go.Figure(data=go.Choropleth(
    locations=df_start['iso_code'],
    z=df_start['co2_per_capita'],
    text=df_start['country'],
    colorscale='Plasma',
    autocolorscale=False,
    zmin=0,
    zmax=20, # Scala fissa per confronto coerente
    marker_line_color='darkgray',
    marker_line_width=0.5,
    colorbar_title="Tons/Persona"
))

fig.update_layout(
    title_text=f"Emissioni CO₂ Pro Capite ({year_start})",
    geo=dict(
        showframe=False,
        showcoastlines=True,
        projection_type='natural earth'
    ),
    margin={"r":0,"t":50,"l":0,"b":0},
    height=500
)

# --- 2. CONVERTIAMO IN WIDGET ---
# Questo è il passaggio chiave: FigureWidget mantiene il grafico vivo in memoria
f_widget = go.FigureWidget(fig)

# --- 3. FUNZIONE DI AGGIORNAMENTO OTTIMIZZATA ---
def aggiorna_dati(change):
    anno_scelto = change['new']

    # Filtriamo i dati per il nuovo anno
    df_new = df[df['year'] == anno_scelto]

    # Aggiorniamo SOLO i dati dentro il grafico esistente (non ricreiamo il grafico)
    with f_widget.batch_update():
        f_widget.data[0].locations = df_new['iso_code']
        f_widget.data[0].z = df_new['co2_per_capita']
        f_widget.data[0].text = df_new['country']
        f_widget.layout.title.text = f"Emissioni CO₂ Pro Capite ({anno_scelto})"

# --- 4. CREAZIONE SLIDER E VISUALIZZAZIONE ---
slider = widgets.IntSlider(
    min=1950,
    max=int(df['year'].max()),
    step=1,
    value=year_start,
    description='Anno:',
    continuous_update=False # Importante per non impallare il browser
)

# Colleghiamo lo slider alla funzione
slider.observe(aggiorna_dati, names='value')

# Visualizziamo tutto insieme
widgets.VBox([slider, f_widget])
```

```python
# 1. ABILITIAMO I WIDGET (Passaggio Fondamentale per Colab)
from google.colab import output
output.enable_custom_widget_manager()

# 2. Importiamo le librerie
import pandas as pd
import plotly.graph_objects as go
from ipywidgets import widgets

# 3. Carichiamo i dati
print("Scaricamento dati...")
url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
df = pd.read_csv(url)
df = df[['iso_code', 'country', 'year', 'co2_per_capita']]
df = df[df['iso_code'].notna()]
print("Dati pronti.")

# 4. Creiamo la Mappa (FigureWidget)
year_start = 2022
df_start = df[df['year'] == year_start]

fig = go.Figure(data=go.Choropleth(
    locations=df_start['iso_code'],
    z=df_start['co2_per_capita'],
    text=df_start['country'],
    colorscale='Plasma',
    zmin=0, zmax=20,
    colorbar_title="Tons/Persona"
))

fig.update_layout(
    title_text=f"Emissioni CO₂ ({year_start})",
    geo=dict(showframe=False, projection_type='natural earth'),
    margin={"r":0,"t":50,"l":0,"b":0},
    height=500
)

# Trasformiamo la figura in un Widget interattivo
g = go.FigureWidget(fig)

# 5. Funzione che aggiorna la mappa quando muovi lo slider
def risposta_slider(change):
    anno = change['new']
    df_new = df[df['year'] == anno]

    # Aggiorna i dati senza ridisegnare tutto (Batch Update)
    with g.batch_update():
        g.data[0].locations = df_new['iso_code']
        g.data[0].z = df_new['co2_per_capita']
        g.data[0].text = df_new['country']
        g.layout.title.text = f"Emissioni CO₂ ({anno})"

# 6. Creiamo lo slider e mostriamo tutto
slider = widgets.IntSlider(
    value=year_start,
    min=1950,
    max=int(df['year'].max()),
    description="Anno:"
)
slider.observe(risposta_slider, names='value')

# Visualizza
widgets.VBox([slider, g])
```
